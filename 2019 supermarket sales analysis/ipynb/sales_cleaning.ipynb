```python
data['Date'] =pd.to_datetime(data['Date'])
#Changing the data type of date so that we can be able to extract year, month and day from the data easily
```

# *Removing whitespaces in strings*

```python
data = data.apply(lambda col: col.str.strip() if pd.api.types.is_string_dtype(col) else col)
```

## *Formatting Dates*

```python
data['Date'] = pd.to_datetime(data['Date'], errors = 'coerce')
data['Date'] = data['Date'].dt.strftime('%Y-%m-%d')
```

# *Removing useless columns*

```python
data.drop(columns=["gross margin percentage"], inplace=True)
```

# *Fix the time column*

```python
def parse_time_to_seconds(time_str):
    parts = str(time_str).split(':')
    minutes = int(parts[0])
    seconds = int(parts[1])
    milliseconds = int(parts[2])
    total_seconds = minutes * 60 + seconds + milliseconds / 1000
    return total_seconds

data['time_sec'] = data['Time'].apply(parse_time_to_seconds)
```

```python
data.drop(columns = 'Time', inplace = True)
```

```python
data.dtypes
```

## *Creating the day and month columns*

```python
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')

# Now extract day and month
data['day'] = data['Date'].dt.day_name()
data['month'] = data['Date'].dt.month_name()
```
