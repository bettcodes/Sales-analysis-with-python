# **<u>EDA Sales Analysis</u>**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("sales_cleaned.csv")
df.head(20)
```

```python
month_order = ["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
df["month"] = pd.Categorical(df["month"], categories=month_order, ordered=True)
df = df.sort_values("month")
```

```python
df.groupby('month')['Total'].sum().plot(kind = "bar")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Total Sales by Month")
plt.show()
```

```python
df.groupby('Branch')['Total'].sum().plot(kind = 'bar')
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.title("Total Sales by Branch")
plt.show()
```

```python
df.groupby('Product line')['Total'].sum().plot(kind = 'bar')
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.title("Total Sales by Product Line")
plt.show()
```

```python
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
df["day"] = pd.Categorical(df["day"], categories=day_order, ordered=True)
df.groupby('day')['Total'].sum().plot(kind = 'bar')
plt.xlabel("Day")
plt.ylabel("Total Sales")
plt.title("Total Sales by Day")
plt.show()
```

```python
df.groupby('City')['Total'].sum().plot(kind = 'bar')
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.title("Total Sales by City")
plt.show()
```

```python
df['Payment'].value_counts().plot(kind = 'pie', autopct='%1.1f%%')
plt.title("Payment Method Distribution")
plt.show()
```

```python
df.groupby('Customer type')['time_sec'].mean().plot(kind = 'bar')
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.barplot(x = df['Product line'], y = df['Total'], hue = df['Gender'])
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.barplot(x = df['month'], y = df['Total'], hue = df['Branch'])
plt.xlabel("Month sales per branch")
plt.ylabel('Sales')
plt.title("Monthly Sales per Branch")
plt.show()
```
